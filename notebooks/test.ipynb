| Parameter | Default value | Description |
|---|---:|---|
| `lasso_lambda` | `1e-3` | Strength of the L1 sparsity penalty. Larger values generally produce fewer nonzero coefficients. |
| `lasso_max_iter` | `500` | Maximum number of FISTA iterations for each LASSO solve. |
| `lasso_tol` | `1e-8` | Relative convergence tolerance for changes in the coefficient vector. |
| `ridge` | `1e-10` | Diagonal regularization added for numerical stability. Fixed internally. |
| `support_tol` | `1e-14` | Minimum coefficient magnitude included in the active support during debiasing. Fixed internally. |
| Convergence stabilizer | `1e-12` | Small constant added to the relative convergence criterion to avoid instability near zero. Fixed internally. |
